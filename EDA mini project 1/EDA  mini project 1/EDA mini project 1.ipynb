```python
# Q13. What is the shape of the distribution of this dataset? Create an appropriate graph to determine that. Take 100 random samples with replacement from this dataset of size 5 each. Create a sampling distribution ofthe mean age of customers. Compare with other sampling distributions of
#sample size 10, 15, 20, 25, 30. State your observations. Does it corroborate the Central Limit Theorem?

# Given data
ages = [42, 44, 62, 35, 20, 30, 56, 20, 23, 41, 55, 22, 31, 27, 66, 21, 18, 24, 42, 25, 32, 50, 31, 26, 36, 39, 40, 18, 36, 22]

# Visualize the original dataset
plt.hist(ages, bins=15, color='blue', edgecolor='black', alpha=0.7)
plt.title('Histogram of Customer Ages (Original Dataset)')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()

# Function to generate sampling distributions
def generate_sampling_distribution(sample_size, num_samples=100):
    sampling_means = []

    for _ in range(num_samples):
        sample = np.random.choice(ages, size=sample_size, replace=True)
        sample_mean = np.mean(sample)
        sampling_means.append(sample_mean)

    return sampling_means

# Generate sampling distributions for different sample sizes
sample_sizes = [5, 10, 15, 20, 25, 30]
plt.figure(figsize=(12, 8))

for sample_size in sample_sizes:
    sampling_means = generate_sampling_distribution(sample_size)
    plt.hist(sampling_means, bins=15, alpha=0.5, label=f'Sample Size {sample_size}')

plt.title('Sampling Distributions of Sample Means')
plt.xlabel('Mean Age')
plt.ylabel('Frequency')
plt.legend()
plt.show()
```

```python
# Q14. Treat this dataset as a binomial distribution where p is the probability that a person ordering an EV is above 50 years age. What is the probability that out of a random sample of 10 buyers exactly 6 are above 50 years of age?
from math import comb
# given probanility
p = sum(x> 50 for x in ages) / len(ages)

# Number of trials (sample size)
n = 10

# Number of successes
k = 6

# Binomial probability mass function
binomianl_pmf = comb(n,k)*(p**k) * ((1-p)**(n-k))
# print the result
print(f'Probability of exactly 6 buyers above 50 years of ages: {binomianl_pmf:.4f}')
```

```python
# Q15. A study claims that 10% of all customers for an EV scooter are above 50 years of age. Using the Normal approximation of a Binomial distribution, find the probability that in a random sample of 300 prospective customers exactly 25 will be above 50 years of age.
#[ Note that the normal distribution can be used to approximate a binomial distribution if np>=5 and nq>=5 with the following correction for continuity P(X=z) = P(z-0.5 < X < z+0.5 ) ]

from scipy.stats  import norm
p = 0.10
n = 300
k= 25
mean = n*p
std_deviation = (n*p*(1-p))** 0.5
z_lower = (k-0.5 - mean) / std_deviation
z_upper = (k+0.5 - mean)/ std_deviation

probability = norm.cdf(z_upper) - norm.cdf(z_lower)
print(f'probability that exactly 25 customers are above 50 years of ages:{probability:.4f}')
```

```python
# Q16. Compute a 95% Confidence Interval for the true mean age of the population of EV scooter buyers for the dataset using appropriate distribution.(State reasons as to why did you use a z or t distribution)

# Given data
ages = [42, 44, 62, 35, 20, 30, 56, 20, 23, 41, 55, 22, 31, 27, 66, 21, 18, 24, 42, 25, 32, 50, 31, 26, 36, 39, 40, 18, 36, 22]

# Sample statistics
sample_mean = statistics.mean(ages)
sample_std_dev = statistics.stdev(ages)
sample_size = len(ages)

# Confidence level
confidence_level = 0.95

# Degrees of freedom for t-distribution
degrees_of_freedom = sample_size - 1

# Calculate the critical value from t-distribution
t_critical = stats.t.ppf((1 + confidence_level) / 2, degrees_of_freedom)

# Calculate the margin of error
margin_of_error = t_critical * (sample_std_dev / (sample_size ** 0.5))

# Calculate the confidence interval
confidence_interval = (sample_mean - margin_of_error, sample_mean + margin_of_error)

# Print the result
print(f"95% Confidence Interval for the true mean age: {confidence_interval}")
```

```python
# Q19. A researcher claims that currently 20% of the population are owning EVs. Test his claim with an alpha =0.05 if out of a random sample of 30 two-wheeler owners only 5 own an EV.


population_proportion_claimed = 0.20
sample_proportion = 5/30
sample_size = 30
alpha = 0.05

standard_error = (population_proportion_claimed*(1-population_proportion_claimed)/ sample_size)**0.5

z_score = (sample_proportion - population_proportion_claimed)/ standard_error
p_value = 2*(1- stats.norm.cdf(abs(z_score)))
reject_null_hypothesis = p_value < alpha

print(f'Z-score: {z_score:.4f}')
print(f'P-value: {p_value:.4f}')
print(f"Null hypothesis rejected? {'Yes' if reject_null_hypothesis else 'No'}")
```

```python
# Q21. Write a report comparing and contrasting Descriptive vs Inferential Statistics in about 200-500 words and how each helps to get more insight into the data at hand

.Descriptive Statistics vs. Inferential Statistics: Unveiling Insights into Data

Statistics is a powerful tool in extracting valuable insights from data. Two main 
branches of statistics, descriptive and inferential statistics, play distinct yet
complementary roles in the analysis process.

.Descriptive Statistics: Painting a Picture of Data

Descriptive statistics focus on summarizing and presenting the main features of a 
dataset. This branch is all about simplifying and organizing raw data in a meaningful
way. Measures such as mean, median, mode, and standard deviation provide a concise summary
of central tendencies and dispersions within a dataset. Visualization tools like histograms,
box plots, and scatter plots are also part of descriptive statistics, offering intuitive
representations of data distributions.

The primary goal of descriptive statistics is to transform complex datasets into manageable information,
making it easier for analysts and decision-makers to grasp the essential characteristics of the data. 
It aids in identifying patterns, trends, and outliers, fostering a preliminary understanding of the 
data's structure and properties.
Inferential Statistics: Drawing Conclusions and Making Predictions
While descriptive statistics reveal the story within a dataset, inferential statistics
take the narrative further by making inferences and predictions about a population based on a sample.
This branch extrapolates findings beyond the observed data, allowing researchers to draw meaningful 
conclusions and generalize their results.Inferential statistics employs probability theory and hypothesis
testing to provide a degree of certainty about the validity of conclusions. Confidence intervals, hypothesis 
tests, and regression analyses are common tools used in this realm. By examining relationships between variables
and assessing the likelihood of observed outcomes, inferential statistics bridge the gap between the sample and 
the broader population.


.Contrasting Descriptive and Inferential Statistics:

1)Scope:

Descriptive Statistics: Focuses on summarizing and visualizing observed data.
Inferential Statistics: Extends findings from a sample to make predictions or draw conclusions about a larger population.

2)Purpose:

Descriptive Statistics: Aims to describe and simplify data, providing an overview.
Inferential Statistics: Aims to generalize findings and make predictions beyond the observed sample.

3)Analysis Techniques:

Descriptive Statistics: Utilizes measures of central tendency, dispersion, and visualization tools.
Inferential Statistics: Involves hypothesis testing, regression analysis, and probability distributions.

4)Application:

Descriptive Statistics: Initial exploration and understanding of the data.
Inferential Statistics: Making predictions, drawing conclusions, and testing hypotheses.
    
    
Role in Gaining Insights:

Descriptive statistics serve as the starting point for any analysis, providing a foundational 
understanding of the dataset's characteristics. It unveils the story behind the numbers, making 
the data more interpretable. However, descriptive statistics alone may fall short in providing a
comprehensive understanding of the broader population.

Enter inferential statistics, which leverages patterns identified through descriptive statistics
to make informed predictions about the population. It adds a layer of depth to the analysis, allowing
researchers to make more robust decisions and draw reliable conclusions.

In conclusion, the synergy between descriptive and inferential statistics is crucial for gaining holistic
insights into data. Descriptive statistics paint the initial picture, while inferential statistics extend
that picture to encompass the entire canvas, unlocking the potential for deeper understanding and informed
decision-making. Both branches are integral to the statistical toolkit, working hand-in-hand to extract
meaningful knowledge from data.






```
